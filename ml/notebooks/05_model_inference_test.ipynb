```python
import joblib

import sys
sys.path.append("..")
from src.preprocessing import preprocess_text
```

```python
model = joblib.load(
    "../models/logistic_regression_model_v1.pkl"
)

tfidf = joblib.load(
    "../models/tfidf.pkl"
)
```

```python
text = """
Aplikasi sekarang makin bagus,
fiturnya keren,
tapi kadang masih gagal login.
"""
```

```python
clean_text = preprocess_text(text)

print(clean_text)
```

```python
vectorized_text = tfidf.transform(
    [clean_text]
)
```

```python
prediction = model.predict(
    vectorized_text
)

print(prediction)
```
