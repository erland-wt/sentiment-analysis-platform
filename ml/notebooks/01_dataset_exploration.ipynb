```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/X_review_GP/raw/X_reviews.csv")
print("\nKolom dalam dataset")
print(df.columns)
print("\n5 Dataset Pertama")
display(df.head())
```

```python
# Missing values
print("\nMissing Values")
print("=" * 30)
print(df.isnull().sum())

# Jumlah data
print("\nJumlah baris & kolom")
print("=" * 30)
print(df.shape)
```

```python
# Distribusi rating
df["Rating"].value_counts()
```

## Yang dipakai hanya kolom "Ulasan" dan "Rating"

```python
# Dataframe bersih
df = df[["Ulasan", "Rating"]]
```

```python
# Hapus missing values
df = df.dropna()
```

```python
# Data Duplikat
print("\nData duplikat")
print(df.duplicated().sum())

# Hapus data duplikat
df = df.drop_duplicates()

print(f"\nData duplikat setelah dihapus: {df.duplicated().sum()}")

# Jumlah data setelah dihapus
print(f"\nJumlah Kolom dan Baris setelah dibersihkan: {df.shape}")

# Distribusi Rating
print(f"\nDistribusi Rating")
print(df["Rating"].value_counts())
```

## Simpan dataset

```python
df.to_csv("../datasets/X_review_GP/interim/cleaned_X_reviews.csv", index=False)
```
